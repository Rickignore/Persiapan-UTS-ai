# Preprocessing

Sebelum dilakukan Klasifikasi -> diperlukan preprocessing data <br>

- Binarzation / Binarisasi
- Mean Removal / Menghilangkan rata-rata
- Scalling
- Normalization

```python
import numpy as np
from sklearn import preprocessing
```

```python
input_data = np.array([[5.1, -2.9, 3.3],
                       [-1.2, 7.8, -6.1],
                       [3.9, 0.4, 2.1],
                       [7.3, -9.9, -4.5]]) # matriks 4 x 3
```

# Binarization

adalah proses mengubah nilai numerik menjadi nilai boolean

```python
data_binarized = preprocessing.Binarizer(threshold=2.1).transform(input_data)
print("Hasil Binarisasi data :\n", data_binarized)
#Kalau nilainya lebih dari 2.1, berubah jadi 1, jika sebaliknya, jadi 0
```

# Mean Removal

ini merupakan teknik preprocessing yang paling digunakan di machine learning. Jadi proses ini menghilangkan rata-rata sehingga hasil rata-ratanya mendekati 0

```python
#Mengetahui Mean dan Standar Deviasi
print("SEBELUM:")
print("Mean = ",input_data.mean(axis=0))
print("Std deviation = ",input_data.std(axis=0))
```

```python
#hasil Mean Removal
hasil_data = preprocessing.scale(input_data)
print("SETELAH:")
print("Mean = ",hasil_data.mean(axis=0))
print("Std deviation =",hasil_data.std(axis=0))
```

```python
print(hasil_data)
```

# Scaling

Preprocessing yang mengubah nilai data, menjadi nilai data baru dengan range antara 0 sampai 1

```python
print(input_data)
```

```python
# Min Max Scaling
data_scaled = preprocessing.MinMaxScaler(feature_range=(0,1))
data_scaled = data_scaled.fit_transform(input_data)
print("Hasil = \n", data_scaled)
```

# Normalization

```python
#L1 normalization (Least Absolute Deviation)
data_normalized_l1 = preprocessing.normalize(input_data, norm='l1')
#L2 normalization (Least Squares)
data_normalized_l2 = preprocessing.normalize(input_data, norm='l2')

print("Hasil normalisasi data menggunakan L1 = \n", data_normalized_l1)
print("\nHasil normalisasi data menggunakan L2 = \n", data_normalized_l2)
```

Normalisasi L1 lebih kokoh daripada Normalisasi L2, karena lebih tahan/resistant terhadap outlier <br>
Jika ingin mempertahankan outliernya, gunakan L2 Normalization

# Label Encoding

Dalam klasifikasi, biasanya sering menghadapi beberapa label. Label dapat berupa kata, nilai, atau yang lain. AI, ML, DL itu selalu membutuhkan angka, sehingga label yang tidak berupa angka, harus diubah menjadi angka (Encoding)

```python
import numpy as np
from sklearn import preprocessing
```

```python
#contoh label
input_label = ['merah', 'hitam', 'merah', 'hijau', 'hitam', 'kuning', 'putih']
```

```python
#Buat label encoder (ditraining)
encoder = preprocessing.LabelEncoder()
encoder.fit(input_label)
```

```python
#print hasilnya
print("Label Mapping = \n")
for i, item in enumerate(encoder.classes_):
  print(item, '-->', i)
```

```python
#Encoder dari label ke angka
test_label = ['hijau', 'merah', 'hitam']
hasil_nilai = encoder.transform(test_label)
print("Label = ",test_label)
print("Hasil Encoder = ", list(hasil_nilai))
```

```python
#Decoder dari angka ke label
test_nilai = [0, 3, 1, 4, 2]
hasil_label = encoder.inverse_transform(test_nilai)
print("Nilai = ",test_nilai)
print("Hasil Decoder = ", list(hasil_label))
```
