Supervised Learning (Classification)

Supervised Learning adalah pembelajaran mesin dimana modelnya dilatih menggunakan data yang berlabel

# Scikit Learn Dataset

```python
from sklearn.datasets import load_iris
iris = load_iris()
#print(iris)

x = iris.data #150 x 4 (baris x kolom) data input
y = iris.target #150 data output
```

```python
print(y)
```

```python
print(x.shape)
print(x)
```

```python
#Train Test Split
#membagi data untuk training 80% dan testing 20% dari total 150 data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state = 1)
```

# Logistic Regression

Logistic Regression merupakan Algoritma pembelajaran yang biasanya digunakan untuk klasifikasi biner menggunakan fungsi logistic (sigmoid) <br>
<br>

Kelebihan:
- Sederhana dan mudah untuk diimplementasikan serta efisien
- Sangat direkomendasikan jika datanya linear <br>

Kekurangan:
- Kurang direkomendasikan jika datanya non-linear
- Sensitif terhadap outlier
- Fitur yang banyak berkorelasi dapat mempengaruhi performa modelnya

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

#Buat model LR
model1 = LogisticRegression()

#Training Modelnya
model1.fit(x_train,y_train)

#Buat prediksi pada data test
y_pred = model1.predict(x_test) #hasil prediksi

#Hitung akurasi modelnya berdasarkan data test dan data prediksi dan LR
akurasi1 = accuracy_score(y_test, y_pred)
print(f'Hasil Akurasi dari LR adalah = {akurasi1 *100:.2f}%')

#Cek perbandingan data test dengan hasil prediksi LR
print("Label sebenarnya", y_test)
print("Label Hasil Prediksi",y_pred)
```

# Naive Bayes

Merupakan algoritma klasifikasi berdasarkan teorema bayes. Algoritma ini dapat digunakan di banyak situasi pada data. <br>
<br>
Kelebihan:
- Waktu training yang dibutuhkan sangat sedikit, bahkan jika datanya banyak
- Sangat efektif untuk klasifikasi berupa teks

Kekurangan:
- Memberikan hasil kurang baik jika datanya tidak seimbang

```python
from sklearn.naive_bayes import GaussianNB

#Buat model NB
model2 = GaussianNB()

#Training Modelnya
model2.fit(x_train,y_train)

#Buat prediksi pada data test
y_pred = model2.predict(x_test) #hasil prediksi

#Hitung akurasi modelnya berdasarkan data test dan data prediksi dan NB
akurasi2 = accuracy_score(y_test, y_pred)
print(f'Hasil Akurasi dari NB adalah = {akurasi2 *100:.2f}%')

#Cek perbandingan data test dengan hasil prediksi NB
print("Label sebenarnya", y_test)
print("Label Hasil Prediksi",y_pred)
```

# Support Vector Machine

Merupakan algoritma machine learning, hal ini sangat baik jika menemukan batas ruang multi dimensi pada data input

```python
from sklearn.svm import SVC #(Support Vector Classification)

#Buat model SVC
model3 = SVC()

#Training Modelnya
model3.fit(x_train,y_train)

#Buat prediksi pada data test
y_pred = model3.predict(x_test) #hasil prediksi

#Hitung akurasi modelnya berdasarkan data test dan data prediksi dan SVC
akurasi3 = accuracy_score(y_test, y_pred)
print(f'Hasil Akurasi dari SVC adalah = {akurasi3 *100:.2f}%')

#Cek perbandingan data test dengan hasil prediksi SVC
print("Label sebenarnya", y_test)
print("Label Hasil Prediksi", y_pred)
```

# Confusion Matrix

Confusion Matrix merupakan tabel yang digunakan untuk mengukur kinerja model klasifikasi. <br>
Confusion Matrix ada 4 Elemen:
- True Positive (TP) = Hasil prediksi memberikan positif, hasil sebenarnya adalah positif juga (Prediksi = "Ini Kucing!", hasil sebenarnya = "Benar, ini kucing")
- True Negative (TN) = Hasil prediksi memberikan negatif, hasil sebenarnya adalah negatif juga (Prediksi = "Ini bukan Kucing!", hasil sebenarnya = "Benar, ini bukan kucing")
- False Positive (FP) = Hasil prediksi memberikan positif, hasil sebenarnya adalah negatif (Prediksi = "Ini Kucing!", hasil sebenarnya = "Salah, ini bukan kucing")
- False Negative (FN) = Hasil prediksi memberikan negatif, hasil sebenarnya adalah positif (Prediksi = "Ini bukan Kucing!", hasil sebenarnya = "Salah, ini kucing")

```python
import seaborn as sns
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
```

```python
true_label = [2, 0, 0, 2, 4, 4, 1, 0, 3, 3, 3]
pred_label = [2, 1, 0, 2, 4, 3, 1, 0, 1, 3, 3]

#buat confussion matrix
confusion_mat = confusion_matrix(true_label, pred_label)

#visualisasi
sns.heatmap(confusion_mat, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Label Prediksi')
plt.ylabel('Label True')
plt.show()
```

```python
#buat confussion matrix
confusion_mat = confusion_matrix(y_test, y_pred)

#visualisasi
sns.heatmap(confusion_mat, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Label Prediksi')
plt.ylabel('Label True')
plt.show()
```
